```python
# ============================================================
# Dataset Comparison Notebook
# Compare transformed_products and cleaned_refurbished_laptops
# ============================================================

from pyspark.sql.functions import avg, max, min, col

# ------------------------------------------------------------
# Load Tables
# ------------------------------------------------------------

products = spark.table("workspace.default.transformed_products")
refurbished = spark.table("workspace.default.cleaned_refurbished_laptops")

print("Datasets Loaded Successfully")

# ------------------------------------------------------------
# Display First 5 Rows
# ------------------------------------------------------------

print("Products Dataset")
products.show(5)

print("Refurbished Dataset")
refurbished.show(5)

# ------------------------------------------------------------
# Print Schema
# ------------------------------------------------------------

print("Products Schema")
products.printSchema()

print("Refurbished Schema")
refurbished.printSchema()

# ------------------------------------------------------------
# Compare Total Rows
# ------------------------------------------------------------

print("Total Rows Comparison")

print("Products Dataset :", products.count())
print("Refurbished Dataset :", refurbished.count())

# ------------------------------------------------------------
# Compare Total Columns
# ------------------------------------------------------------

print("Total Columns Comparison")

print("Products Dataset :", len(products.columns))
print("Refurbished Dataset :", len(refurbished.columns))

# ------------------------------------------------------------
# Compare Average Price
# ------------------------------------------------------------

print("Average Price Comparison")

products.select(
    avg("Price").alias("Average_Product_Price")
).show()

refurbished.select(
    avg("Price_euro").alias("Average_Refurbished_Price")
).show()

# ------------------------------------------------------------
# Compare Maximum Price
# ------------------------------------------------------------

print("Maximum Price Comparison")

products.select(
    max("Price").alias("Maximum_Product_Price")
).show()

refurbished.select(
    max("Price_euro").alias("Maximum_Refurbished_Price")
).show()

# ------------------------------------------------------------
# Compare Minimum Price
# ------------------------------------------------------------

print("Minimum Price Comparison")

products.select(
    min("Price").alias("Minimum_Product_Price")
).show()

refurbished.select(
    min("Price_euro").alias("Minimum_Refurbished_Price")
).show()

# ------------------------------------------------------------
# RAM vs Memory Comparison
# ------------------------------------------------------------

print("RAM Distribution (Products)")

display(
    products.groupBy("RAM_Size")
            .count()
            .orderBy(col("RAM_Size"))
)

print("Memory Distribution (Refurbished)")

display(
    refurbished.groupBy("Memory_GB")
                .count()
                .orderBy(col("Memory_GB"))
)

# ------------------------------------------------------------
# SSD vs Storage Comparison
# ------------------------------------------------------------

print("SSD Distribution (Products)")

display(
    products.groupBy("SSD_Size")
            .count()
            .orderBy(col("SSD_Size"))
)

print("Storage Distribution (Refurbished)")

display(
    refurbished.groupBy("Storage_GB")
                .count()
                .orderBy(col("Storage_GB"))
)

# ------------------------------------------------------------
# Operating System Distribution
# ------------------------------------------------------------

print("Operating System Distribution")

display(
    products.groupBy("OS")
            .count()
            .orderBy(col("count").desc())
)

# ------------------------------------------------------------
# Refurbished Grade Distribution
# ------------------------------------------------------------

print("Grade Distribution")

display(
    refurbished.groupBy("Grade")
                .count()
                .orderBy(col("count").desc())
)

# ------------------------------------------------------------
# Refurbished Storage Type Distribution
# ------------------------------------------------------------

print("Storage Type Distribution")

display(
    refurbished.groupBy("Storage_type")
                .count()
                .orderBy(col("count").desc())
)

# ------------------------------------------------------------
# Refurbished Memory Type Distribution
# ------------------------------------------------------------

print("Memory Type Distribution")

display(
    refurbished.groupBy("Memory_type")
                .count()
                .orderBy(col("count").desc())
)

# ------------------------------------------------------------
# Compare Top 10 Expensive Products
# ------------------------------------------------------------

print("Top 10 Expensive New Laptops")

display(
    products.select("Model","Price")
            .orderBy(col("Price").desc())
            .limit(10)
)

print("Top 10 Expensive Refurbished Laptops")

display(
    refurbished.select(
        "ID",
        "Model_year",
        "Price_euro"
    )
    .orderBy(col("Price_euro").desc())
    .limit(10)
)

# ------------------------------------------------------------
# Compare Cheapest Products
# ------------------------------------------------------------

print("Top 10 Cheapest New Laptops")

display(
    products.select("Model","Price")
            .orderBy(col("Price"))
            .limit(10)
)

print("Top 10 Cheapest Refurbished Laptops")

display(
    refurbished.select(
        "ID",
        "Model_year",
        "Price_euro"
    )
    .orderBy(col("Price_euro"))
    .limit(10)
)

# ------------------------------------------------------------
# Dataset Summary
# ------------------------------------------------------------

print("Comparison Summary")

print("---------------------------------------")
print("Products Rows :", products.count())
print("Refurbished Rows :", refurbished.count())

print("Products Columns :", len(products.columns))
print("Refurbished Columns :", len(refurbished.columns))
print("---------------------------------------")

print("Dataset Comparison Completed Successfully!")
```
