```python
# Import required functions

from pyspark.sql.functions import avg, max, min, col, count


# Load refurbished laptop table

df = spark.table("cleaned_refurbished_laptops")


# Show first 5 rows

df.show(5)


# Print schema

df.printSchema()


# Total rows and columns

print("Total Rows:", df.count())
print("Total Columns:", len(df.columns))


# Summary statistics

display(df.describe())


# Display selected columns

display(
    df.select(
        "ID",
        "Model_year",
        "Memory_GB",
        "Storage_GB",
        "Storage_type",
        "Memory_type",
        "Grade",
        "Price_euro"
    )
)


# Grade distribution

display(
    df.groupBy("Grade")
      .count()
      .orderBy(col("count").desc())
)


# Storage type distribution

display(
    df.groupBy("Storage_type")
      .count()
      .orderBy(col("count").desc())
)


# Memory type distribution

display(
    df.groupBy("Memory_type")
      .count()
      .orderBy(col("count").desc())
)


# RAM / Memory distribution

display(
    df.groupBy("Memory_GB")
      .count()
      .orderBy(col("Memory_GB"))
)


# Storage size distribution

display(
    df.groupBy("Storage_GB")
      .count()
      .orderBy(col("Storage_GB"))
)


# Touch screen distribution

display(
    df.groupBy("Touch_screen")
      .count()
      .orderBy(col("count").desc())
)


# Model year distribution

display(
    df.groupBy("Model_year")
      .count()
      .orderBy(col("Model_year"))
)


# Price statistics

df.select(
    avg("Price_euro").alias("Average_Refurbished_Price"),
    max("Price_euro").alias("Maximum_Refurbished_Price"),
    min("Price_euro").alias("Minimum_Refurbished_Price")
).show()


# Top 10 expensive refurbished laptops

display(
    df.select(
        "ID",
        "Model_year",
        "Memory_GB",
        "Storage_GB",
        "Grade",
        "Price_euro"
    )
    .orderBy(
        col("Price_euro").desc()
    )
    .limit(10)
)


# Top 10 cheapest refurbished laptops

display(
    df.select(
        "ID",
        "Model_year",
        "Memory_GB",
        "Storage_GB",
        "Grade",
        "Price_euro"
    )
    .orderBy(
        col("Price_euro")
    )
    .limit(10)
)


# Average price by Grade

display(
    df.groupBy("Grade")
      .agg(
          avg("Price_euro").alias("Average_Price")
      )
      .orderBy(
          col("Average_Price").desc()
      )
)


# Average price by Storage Type

display(
    df.groupBy("Storage_type")
      .agg(
          avg("Price_euro").alias("Average_Price")
      )
      .orderBy(
          col("Average_Price").desc()
      )
)


print("Refurbished Laptop Databricks Analysis Completed Successfully!")
```
