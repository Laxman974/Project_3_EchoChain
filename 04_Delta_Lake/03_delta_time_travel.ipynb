```python
# ============================================
# Delta Time Travel
# ============================================

print("History of Delta Table")

display(
    spark.sql(
        "DESCRIBE HISTORY delta_transformed_products"
    )
)

print("Current Version")

display(
    spark.read
    .format("delta")
    .table("delta_transformed_products")
)

print("Version 0")

display(
    spark.read
    .format("delta")
    .option("versionAsOf",0)
    .table("delta_transformed_products")
)

print("Time Travel Completed Successfully!")
```
